# Main workflow

### Create total arrays of features

```python
with h5py.File(file_path, 'r') as hdf:
    # array_scores = np.array([])
    # array_features = np.empty((0, 14))
    for subject in hdf.keys():
        # if "SC" in subject:
        array_scores = np.array([])
        array_features = np.empty((0, 14))
        #array_features = np.empty((0, 10))
        group = hdf[subject]
        print(group.keys())
        array_scores = np.concatenate((array_scores, group['Mapped_scores'][:]), axis = 0)
        #array_scores = np.concatenate((array_scores, group['sleep_states'][:]), axis = 0)
        array_features = np.concatenate((array_features, group['Features'][:]), axis = 0)
        #array_features = np.concatenate((array_features, group['features'][:]), axis = 0)
        #array_features = np.concatenate((array_features, group['Features'][:, :10]), axis = 0)
        # indices_artefact = np.where(array_scores == 5)[0] # Artefact
        indices_wake = np.where(array_scores == 0)[0] # Wake
        indices_N1 = np.where(array_scores == 1)[0] # N1
        indices_N2 = np.where(array_scores  == 2)[0] # N2
        indices_N3 = np.where(array_scores == 3)[0] # N3
        indices_rem = np.where(array_scores == 4)[0] # REM
        len_scores = len(array_scores_total)

        array_scores_total = np.concatenate((array_scores_total, array_scores), axis = 0)
        array_features_total = np.concatenate((array_features_total, array_features), axis = 0)
        # indices_artefact = np.array([el + len_scores for el in indices_artefact])
        indices_wake = np.array([el + len_scores for el in indices_wake])
        indices_N1 = np.array([el + len_scores for el in indices_N1])
        indices_N2 = np.array([el + len_scores for el in indices_N2])
        indices_N3 = np.array([el + len_scores for el in indices_N3])
        indices_rem = np.array([el + len_scores for el in indices_rem])

        # indices_artefact_total = np.concatenate((indices_artefact_total, indices_artefact), axis = 0)
        indices_wake_total = np.concatenate((indices_wake_total, indices_wake), axis = 0)
        indices_n1_total = np.concatenate((indices_n1_total, indices_N1), axis = 0)
        indices_n2_total = np.concatenate((indices_n2_total, indices_N2), axis = 0)
        indices_n3_total = np.concatenate((indices_n3_total, indices_N3), axis = 0)
        indices_rem_total = np.concatenate((indices_rem_total, indices_rem), axis = 0)
        # else:
        #     continue
        
states = [el.lower() for el in states]

# if "artefact" in states:
#     features_artefact = array_features_total[indices_artefact_total.astype(int)]
#     scores = array_scores_total[indices_artefact_total.astype(int)]
#     dataset = np.concatenate((dataset, features_artefact), axis=0)
#     manual_scoring = np.concatenate((manual_scoring, scores), axis=0)

if "wake" in states:
    features_wake = array_features_total[indices_wake_total.astype(int)]
    scores = array_scores_total[indices_wake_total.astype(int)]
    dataset = np.concatenate((dataset, features_wake), axis=0)
    manual_scoring = np.concatenate((manual_scoring, scores), axis=0)

if "n1" in states:
    features_n1 = array_features_total[indices_n1_total.astype(int)]
    scores = array_scores_total[indices_n1_total.astype(int)]
    dataset = np.concatenate((dataset, features_n1), axis=0)
    manual_scoring = np.concatenate((manual_scoring, scores), axis=0)

if "n2" in states:
    features_n2 = array_features_total[indices_n2_total.astype(int)]
    scores = array_scores_total[indices_n2_total.astype(int)]
    dataset = np.concatenate((dataset, features_n2), axis=0)
    manual_scoring = np.concatenate((manual_scoring, scores), axis=0)

if "n3" in states:
    features_n3 = array_features_total[indices_n3_total.astype(int)]
    scores = array_scores_total[indices_n3_total.astype(int)]
    dataset = np.concatenate((dataset, features_n3), axis=0)
    manual_scoring = np.concatenate((manual_scoring, scores), axis=0)

if "rem" in states:
    features_rem = array_features_total[indices_rem_total.astype(int)]
    scores = array_scores_total[indices_rem_total.astype(int)]
    dataset = np.concatenate((dataset, features_rem), axis=0)
    manual_scoring = np.concatenate((manual_scoring, scores), axis=0)

dataset = np.array(dataset)
manual_scoring = np.array(manual_scoring).astype(int)
```

## Normal dataset

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    dataset,
    manual_scoring,
    test_size=0.25,
    stratify=manual_scoring,
    random_state=42
)
```

```python
# This code was generated with ChatGPT, but manually verified and edited
# prompt: "Create a print to display the count and ratios of the states 
#          compared to the total data size, sorted by state"

import numpy as np
from collections import Counter

def compute_state_distro(states, title_str):
    total = len(states)

    counts = Counter(states)

    with open(os.path.join(output_path, f"{dataset_name}_state_distribution.txt"), "a") as f:
        print("-" * 50, file=f)
        print(f"Distribution of sleep states in {title_str} dataset:", file=f)
        print(f"Total epochs: {total}", file=f)

        for state in sorted(counts.keys(), key=lambda x: int(x)):
            state_str = str(state)
            count = counts[state]
            ratio = count / total * 100
            print(f"{state_str:10s}  Count: {count:7d}   Ratio: {ratio:6.2f}%", file=f)
        print("-" * 50, "\n", file=f)


compute_state_distro(manual_scoring, "total")
compute_state_distro(y_train, "training")
compute_state_distro(y_test, "testing")
```

### save dataset and scoring to required filetypes

```python
# save training dataset
np.save(os.path.join(output_path, "training/features.npy"), X_train)
# save training labels
savemat(os.path.join(output_path, "training/states.mat"), {"states": y_train})

# save testing dataset
np.save(os.path.join(output_path, "testing/features.npy"), X_test)
# save testing labels
savemat(os.path.join(output_path, "testing/states.mat"), {"states": y_test})
```
